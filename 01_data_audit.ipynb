## Conflicting Labels Analysis

This section identifies conflicting labels in the dataset by checking whether identical feature combinations are associated with different target labels (`is_fraud`). This helps detect potential inconsistencies in the dataset.

### 1. Selecting Feature Columns

```python
feature_columns = df.columns.drop("is_fraud").tolist()
```

- `df.columns` retrieves all column names from the dataset.
- `.drop("is_fraud")` removes the target variable from the column list.
- `.tolist()` converts the remaining column names into a Python list.
- The resulting list is stored in `feature_columns`.
- This ensures that only input features are considered during the analysis, excluding the target variable.

### 2. Grouping Identical Feature Combinations

```python
feature_groups = (
    df.groupby(feature_columns, dropna=False)
    .agg(
        total_rows=("is_fraud", "size"),
        unique_labels=("is_fraud", "nunique")
    ).reset_index()
)
```

- `df.groupby(feature_columns, dropna=False)` groups records having identical values across all feature columns.
- `dropna=False` ensures that missing values are also considered while grouping records.
- `.agg()` performs multiple aggregation operations on each group.

**Inside `.agg()`:**

- `total_rows=("is_fraud", "size")` counts the total number of records in each feature group.
- `unique_labels=("is_fraud", "nunique")` counts the number of distinct target labels present in each group.
- `.reset_index()` converts the grouped result back into a regular DataFrame.
- The resulting DataFrame is stored in `feature_groups`.

### 3. Identifying Conflicting Label Groups

```python
conflicting_groups = feature_groups[feature_groups["unique_labels"] > 1]
```

- `feature_groups["unique_labels"]` accesses the number of unique target labels in each group.
- `> 1` identifies groups containing more than one distinct target label.
- These groups have identical input features but different `is_fraud` values.
- The filtered records are stored in `conflicting_groups`.

**Example:**

| Transaction Features | is_fraud |
|---|---:|
| Feature combination A | 0 |
| Feature combination A | 1 |

Here, the input features are identical, but the target labels are different. Therefore, this is a conflicting label group.

### 4. Counting Unique Feature Combinations

```python
print("Unique feature combinations: ", len(feature_groups))
```

- `len(feature_groups)` calculates the total number of unique feature combinations.
- Each group represents one distinct combination of input feature values.
- This helps understand how many different transaction patterns exist in the dataset.

### 5. Counting Conflicting Label Groups

```python
print("conflicting label groups: ", len(conflicting_groups))
```

- `len(conflicting_groups)` calculates the total number of groups containing conflicting target labels.
- A conflicting group has identical feature values but more than one target label.
- This helps identify potential inconsistencies in the dataset.

### 6. Counting Rows in Conflicting Groups

```python
print("\nRows in conflicting groups : ", conflicting_groups["total_rows"].sum())
```

- `conflicting_groups["total_rows"]` accesses the total record count of each conflicting group.
- `.sum()` calculates the combined number of rows across all conflicting groups.
- This tells us how many transaction records are involved in label conflicts.

### 7. Displaying the First Ten Conflicting Groups

```python
display(conflicting_groups.head(10))
```

- `.head(10)` retrieves the first ten conflicting groups.
- `display()` presents the results in a readable tabular format.
- This allows us to inspect the feature combinations, total records, and number of unique labels in conflicting groups.

### Objective

The main objective of this analysis is to identify whether identical transaction features have been assigned different fraud labels.

Such inconsistencies may indicate labeling errors or ambiguous records in the dataset. Identifying these cases helps improve data quality and maintain reliable training data for the fraud detection model.

````python
feature_columns = df.columns.drop("is_fraud").tolist()

feature_groups = (
    df.groupby(feature_columns, dropna=False)
    .agg(
        total_rows = ("is_fraud","size"),
        unique_labels=("is_fraud","nunique")
    ).reset_index()
)

conflicting_groups = feature_groups[feature_groups["unique_labels"] > 1]

print("Unique feature combinations: ",len(feature_groups))

print("conflicting label groups: ",len(conflicting_groups))

print("\nRows in conflicting groups : ",conflicting_groups["total_rows"].sum())

display(conflicting_groups.head(10))
````

## Timestamp Data Validation and Chronological Order Analysis

This section performs validation checks on transaction timestamps. It identifies invalid timestamps, finds the earliest and latest transaction dates, and verifies whether transactions are arranged in chronological order.

### 1. Checking Invalid Timestamps

```python id="p8zv3k"
print("Invalid timestamps : ")
print(timestamps.isna().sum())
```

- `.sum()` counts the total number of invalid or missing timestamps.
- `print()` displays the total count.
- This helps determine whether any timestamp values failed during datetime conversion.

### 2. Finding the Earliest Timestamp

```python id="v5h8jk"
print("Earlist :", timestamps.min())
```

- `timestamps.min()` returns the earliest (minimum) datetime value in the timestamp column.
- This identifies the starting point of the transaction timeline.
- It helps understand when the first transaction occurred.

### 3. Finding the Latest Timestamp

```python id="x7k2lm"
print("Latest :", timestamps.max())
```

- `timestamps.max()` returns the latest (maximum) datetime value.
- This identifies the ending point of the transaction timeline.
- It helps determine the overall time range covered by the dataset.

### 4. Checking Chronological Order

```python id="j4n9qw"
print(
    "Already chronological : ",
    timestamps.is_monotonic_increasing
)
```

- `timestamps.is_monotonic_increasing` checks whether timestamps are arranged in non-decreasing order.
- It returns `True` if every timestamp is greater than or equal to the previous timestamp.
- It returns `False` if any timestamp appears before the previous timestamp.
- This helps verify whether transaction records are chronologically arranged.

### 5. Counting Out-of-Order Adjacent Rows

```python id="m2b6zr"
print(
    "Out-of-order adjacent rows:",
    (timestamps.diff() < pd.Timedelta(0)).sum()
)
```

- `timestamps.diff()` calculates the time difference between each timestamp and the previous row's timestamp.
- `pd.Timedelta(0)` represents a zero time difference.
- `<` checks whether the time difference is negative.
- A negative difference indicates that the current timestamp is earlier than the previous timestamp.
- `.sum()` counts the total number of adjacent rows that are out of chronological order.
- This provides a numerical count of chronological ordering violations.

### Objective

The main objective of this section is to verify the validity and chronological consistency of transaction timestamps.

The analysis helps us:

1. Identify invalid or missing timestamps.
2. Determine the earliest transaction date.
3. Determine the latest transaction date.
4. Verify whether transactions are chronologically arranged.
5. Count adjacent records that violate chronological order.

These checks are important for real-time fraud detection because transaction timing plays a significant role in identifying suspicious activities and unusual transaction patterns.
