
# 4. Duplicate Investigation

## 4.1 Objective

The objective of duplicate investigation is to identify
identical records, understand their class distribution,
and determine whether identical input features have
conflicting target labels.

Duplicate analysis is important because repeated records
can affect model evaluation, class distribution, and the
reliability of training and testing results.

No duplicate records are removed during this investigation.

## 4.2 Rows Involved in Duplicate Groups

### Code

```python
duplicate_mask = df.duplicated(keep=False)
```

### Explanation

`df.duplicated()` identifies rows that exactly match an
earlier row across all columns.

By default, `keep="first"` marks the first occurrence as
not duplicated and subsequent occurrences as duplicated.

When `keep=False` is used, every row belonging to a
duplicate group is marked `True`, including the first
occurrence.

The resulting Boolean Series is stored in `duplicate_mask`.

### Result

Rows involved in duplicate groups: 1,854

### Interpretation

A total of 1,854 rows belong to groups containing identical
records. This is different from the number of repeated
occurrences after the first appearance.

## 4.3 Class Distribution in Duplicate Groups

### Code

```python
df.loc[duplicate_mask, "Class"].value_counts()
```

### Explanation

`df.loc[duplicate_mask, "Class"]` selects the `Class`
values of rows where `duplicate_mask` is `True`.

`value_counts()` counts the occurrences of each target label
within the selected rows.

### Result

| Class | Rows |
|---|---:|
| 0 — Legitimate | 1,822 |
| 1 — Fraud | 32 |
| Total | 1,854 |

### Interpretation

The duplicate-group rows contain 1,822 legitimate-labeled
rows and 32 fraud-labeled rows.

These counts represent rows involved in duplicate groups,
not the number of unique transactions.

## 4.4 Repeated Rows After the First Occurrence

### Code

```python
df.duplicated().sum()
```

### Explanation

With the default `keep="first"`, Pandas marks subsequent
occurrences of identical rows as `True`.

Summing the Boolean results counts the repeated occurrences
after the first occurrence in each group.

### Result

Repeated rows after the first occurrence: 1,081

### Interpretation

The dataset contains 1,081 additional occurrences of
records that have already appeared.

This does not mean that exactly 1,081 unique transactions
are erroneous.

## 4.5 Exact Duplicate Groups

### Code

```python
duplicate_groups = (
    df.groupby(df.columns.tolist(), dropna=False)
      .size()
      .reset_index(name="count")
)

duplicate_groups = duplicate_groups[
    duplicate_groups["count"] > 1
]
```

### Explanation

`df.columns.tolist()`

Creates a list containing all column names, including
the target column `Class`.

`groupby(df.columns.tolist(), dropna=False)`

Groups rows by the exact values of every column.

Because `Class` is included in the grouping keys, records
with different target labels belong to different groups.

`dropna=False` retains groups containing missing values,
if any exist.

`.size()`

Counts the number of original rows in each group.

`.reset_index(name="count")`

Converts the grouping keys back into regular columns and
stores the group size in a column named `count`.

The filtering condition:

`duplicate_groups["count"] > 1`

retains only groups that occur more than once.

### Result

Number of exact duplicate groups: 773

### Interpretation

There are 773 distinct full-row patterns that occur more
than once in the original dataset.

Each group has identical values across all 31 columns,
including the target label.

The first ten groups are displayed for inspection.

## 4.6 Class Distribution Across Duplicate Groups

### Code

```python
duplicate_groups.groupby("Class")["count"].sum().astype(int)
```

### Explanation

`groupby("Class")`

Groups the previously identified duplicate groups by
their target label.

`["count"]`

Selects the number of original rows represented by each
duplicate group.

`.sum()`

Adds the original row counts for each class.

`.astype(int)`

Converts the resulting counts to integer data type.

### Result

| Class | Rows represented by duplicate groups |
|---|---:|
| 0 — Legitimate | 1,822 |
| 1 — Fraud | 32 |

### Interpretation

These values count all original rows belonging to repeated
full-row patterns, including their first occurrences.

They are not counts of duplicate groups themselves.

## 4.7 Summary of Findings

- Rows involved in duplicate groups: 1,854
- Repeated occurrences after the first: 1,081
- Exact duplicate groups: 773
- Class 0 rows in duplicate groups: 1,822
- Class 1 rows in duplicate groups: 32

The counts are internally consistent:

1,854 - 1,081 = 773

This works because every duplicate group contributes one
first occurrence, and every additional occurrence is counted
as a repeated row.

## 4.8 Engineering Decision

No records are deleted at this stage.

Identical records do not automatically prove that the data
is invalid. The dataset does not provide a unique transaction
identifier to confirm whether identical records represent
the same real-world event or separate events with identical
recorded values.

Before deciding on a removal strategy, further analysis is
required to understand the effect of duplicates on model
training and evaluation.

## 4.9 Next Investigation

Investigate whether identical input features are associated
with different target labels.

For this analysis, the `Class` column must be excluded from
the grouping keys so that identical inputs with different
labels can be detected.
